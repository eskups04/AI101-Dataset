```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
url = "https://raw.githubusercontent.com/eskups04/AI101-Dataset/main/ai_student_impact_dataset.csv"
df = pd.read_csv(url)

feature_names = [
    "Weekly_GenAI_Hours",
    "Traditional_Study_Hours",
    "Tool_Diversity",
    "Anxiety_Level_During_Exams",
]

X_raw = df[feature_names].to_numpy(dtype=float)
y = df["Skill_Retention_Score"].to_numpy(dtype=float)

assert X_raw is not None and y is not None, "Load your dataset above before continuing."
n_samples, n_features = X_raw.shape
print(f"n_samples = {n_samples}, n_features = {n_features}")
print(f"feature_names = {feature_names}")
```
