**What to do:** Represent your dataset as a design matrix X ∈ ℝⁿˣᵖ. Compute the product XᵀX.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

url = "https://raw.githubusercontent.com/eskups04/AI101-Dataset/main/ai_student_impact_dataset.csv"
df = pd.read_csv(url)

feature_names = [
    "Weekly_GenAI_Hours",
    "Traditional_Study_Hours",
    "Tool_Diversity",
    "Anxiety_Level_During_Exams",
]

X_raw = df[feature_names].to_numpy(dtype=float)
y = df["Skill_Retention_Score"].to_numpy(dtype=float)

print("Loaded successfully! Shape:", X_raw.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
X = np.asarray(X_raw, dtype=float)
print("X shape:", X.shape)
```

```python
XtX = X.T @ X
print("X^T X shape:", None if XtX is None else XtX.shape)
print("\nX^T X matrix:\n", XtX)
```

```python
i, j = 0, 1
print(f"(X^T X)[{i},{j}] = {XtX[i, j]:.4f} -- corresponds to features "
      f"'{feature_names[i]}' and '{feature_names[j]}'")
```
