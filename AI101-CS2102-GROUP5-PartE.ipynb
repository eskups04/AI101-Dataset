**What to do:** Compute det(XᵀX) on your training data (after your Part D transformation) before attempting to invert it.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

url = "https://raw.githubusercontent.com/eskups04/AI101-Dataset/main/ai_student_impact_dataset.csv"
df = pd.read_csv(url)

feature_names = [
    "Weekly_GenAI_Hours",
    "Traditional_Study_Hours",
    "Tool_Diversity",
    "Anxiety_Level_During_Exams",
]

X = df[feature_names].to_numpy(dtype=float)
y = df["Skill_Retention_Score"].to_numpy(dtype=float)

mu = X.mean(axis=0)
sigma = X.std(axis=0)
sigma_safe = np.where(sigma == 0, 1.0, sigma)
S = np.diag(1.0 / sigma_safe)
X_transformed = (X - mu) @ S

rng = np.random.default_rng(42)
n = X_transformed.shape[0]
perm = rng.permutation(n)
train_size = int(0.8 * n)
train_idx, test_idx = perm[:train_size], perm[train_size:]

X_train, X_test = X_transformed[train_idx], X_transformed[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print("Setup complete! X_train shape:", X_train.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
XtX_train = X_train.T @ X_train
det_XtX_train = np.linalg.det(XtX_train)
print(f"det(X_train^T X_train) = {det_XtX_train:.4e}")
```

```python
cond_number = np.linalg.cond(XtX_train)
print(f"Is XtX_train invertible? {not np.isclose(det_XtX_train, 0.0)}")
print(f"Condition number = {cond_number:.4f}")
```
