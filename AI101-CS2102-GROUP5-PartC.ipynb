**What to do:** Compute the rank of X. Determine whether its columns are linearly independent.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

url = "https://raw.githubusercontent.com/eskups04/AI101-Dataset/main/ai_student_impact_dataset.csv"
df = pd.read_csv(url)

feature_names = [
    "Weekly_GenAI_Hours",
    "Traditional_Study_Hours",
    "Tool_Diversity",
    "Anxiety_Level_During_Exams",
]

X_raw = df[feature_names].to_numpy(dtype=float)
X = X_raw.copy()
XtX = X.T @ X
y = df["Skill_Retention_Score"].to_numpy(dtype=float)

print("Loaded directly from GitHub! Shape:", X.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
rank_X = np.linalg.matrix_rank(X)
n_samples, n_features = X.shape
print(f"X is {n_samples} x {n_features}.  rank(X) = {rank_X}")
```

```python
if rank_X == n_features:
    print(f"Since rank(X) = {rank_X} equals the number of features ({n_features}), "
          "X has full column rank and its columns are linearly independent.")
else:
    print(f"Since rank(X) = {rank_X} < {n_features}, X is rank-deficient "
          "and its columns are linearly dependent.")
```
