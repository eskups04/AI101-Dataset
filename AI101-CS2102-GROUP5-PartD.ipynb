```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

url = "https://raw.githubusercontent.com/eskups04/AI101-Dataset/main/ai_student_impact_dataset.csv"
df = pd.read_csv(url)

feature_names = [
    "Weekly_GenAI_Hours",
    "Traditional_Study_Hours",
    "Tool_Diversity",
    "Anxiety_Level_During_Exams",
]

X = df[feature_names].to_numpy(dtype=float)
y = df["Skill_Retention_Score"].to_numpy(dtype=float)

# Define reproducible 80/20 split indices upfront
rng = np.random.default_rng(42)
n = X.shape[0]
perm = rng.permutation(n)
train_size = int(0.8 * n)
train_idx, test_idx = perm[:train_size], perm[train_size:]

print("Loaded X:", X.shape, "| y:", y.shape)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

```python
# Fit centering (mu) and scaling (sigma, S) strictly on training indices
mu = X[train_idx].mean(axis=0)
sigma = X[train_idx].std(axis=0)
sigma_safe = np.where(sigma == 0, 1.0, sigma)
S = np.diag(1.0 / sigma_safe)

X_centered = X - mu
X_transformed = X_centered @ S
print("X_transformed shape:", X_transformed.shape)
```

```python
idx = 0
x_before = X[idx]
x_after = X_transformed[idx]
print("Before:", x_before)
print("After: ", x_after)
```

```python
X_train, X_test = X_transformed[train_idx], X_transformed[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

print("X_train shape:", X_train.shape, "| X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape, "| y_test shape:", y_test.shape)
```

> *Your answer:* _I applied standardization, which consists of mean centering and scaling by the standard deviation of each independent variable. > *Your answer:* _Our standardization transformation is invertible because its scaling matrix is diagonal and has nonzero diagonal entries. Invertibility allows us to convert transformed values back into their original units, making predictions easier to interpret. **Q16.** Would your transformation's matrix change if computed from a different subset of your data (e.g., a different train/test split)? What does that imply about how you must apply it consistently to both training and test data?
